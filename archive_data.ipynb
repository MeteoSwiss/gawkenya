```python
import os
import shutil
import logging

logging.basicConfig(filename='archive_data.log', level='ERROR')

source = "D:/data"
target = "M:/pay-proj/pay/repos_pay/GAW_Kenya/MKN"
os.listdir(source)


for root, dirs, files in os.walk(source):
    for file in files:
        logging.log(f"{root}: {len(files)} files.")
        if not os.path.exists(os.path.join(source, file)):
            try:
                os.makedirs(os.path.join(root).replace(source, target), exist_ok=True)
                shutil.move(os.path.join(source, file), os.path.join(target, file))
            except Exception as err:
                logging.log(err)
```

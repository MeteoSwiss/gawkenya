```python
# Combine all AE31 level1 data
ae31_pattern = "A11a.parquet"
df_ae31_1m = pl.DataFrame()
source = "data/level1/mkn"

for root, dirs, files in os.walk(source):
    for file in files:
        if file in ae31_pattern:
            _path = os.path.join(root, file)
            _df = pl.read_parquet(_path)
            _df = pl_simplify_dtypes(_df)
            _df.write_parquet(_path)
            df_ae31_1m = pl.concat([df_ae31_1m, _df], how='diagonal')
            logger.info(f"Added {len(_df)} rows from {_path} to df_ae31_1m")

df_ae31_1m.write_parquet(os.path.join(source, "ae31_1m_combined.parquet"))

display(df_ae31_1m.describe())
```

```python
# Combine all AE33 level1 data
ae33_pattern = "ae33.parquet"
df_ae33_1m = pl.DataFrame()
source = "data/level1/mkn"

for root, dirs, files in os.walk(source):
    for file in files:
        if file in ae33_pattern:
            _path = os.path.join(root, file)
            _df = pl.read_parquet(_path)
            _df = _df.with_columns(pl.col('dtm').dt.replace_time_zone(time_zone='UTC'))
            _df = pl_simplify_dtypes(_df)
            _df.write_parquet(_path)
            df_ae33_1m = pl.concat([df_ae33_1m, _df], how='diagonal')
            logger.info(f"Added {len(_df)} rows from {_path} to df_ae33_1m")

df_ae33_1m.write_parquet(os.path.join(source, "ae33_1m_combined.parquet"))

display(df_ae33_1m.describe())
```

```python
# Filter records where 'f_BC6' is NOT (0 or null)
df_ae33_1m_filtered = df_ae33_1m.filter(pl.col("f_BC6").is_null() | (pl.col("f_BC6") == '0'))

# Aggregate to hourly values
df_ae33_1h = (
    df_ae33_1m_filtered
    .with_columns(pl.col("dtm").dt.truncate("1h").alias("hour_start"))
    .group_by("hour_start")
    .agg(
        pl.col("BC1").mean(),
        pl.col("BC2").mean(),
        pl.col("BC3").mean(),
        pl.col("BC4").mean(),
        pl.col("BC5").mean(),
        pl.col("BC6").mean(),
        pl.col("BC7").mean(),
    )
    .with_columns(pl.col("hour_start").dt.offset_by("1h"))  # Shift to hour-end
)

# Rename timestamp column to final format
df_ae33_1h = df_ae33_1h.rename({"hour_start": "dtm"})

df_ae33_1h.write_parquet(os.path.join(source.replace('level1', 'level2'), "ae33_1h_combined.parquet"))

# overall statistics on coverage
num_hours_in_period = df_ae33_1h.select((pl.col("dtm").max() - pl.col("dtm").min()).dt.total_hours()).item()
coverage = len(df_ae33_1h) / num_hours_in_period
print(f"Period covered: {df_ae33_1h['dtm'].min()} thru {df_ae33_1h['dtm'].max()}")
print(f"Temporal coverage of 1 hourly data: {coverage}.")
```

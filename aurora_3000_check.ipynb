```python
import polars as pl

# zero check
def print_zero_check_results(df):
    _df = df.filter((pl.col("DateTime").cast(pl.Datetime) >= pl.datetime(2023, 10, 20, 8, 5, 00))
                    & (pl.col("DateTime").cast(pl.Datetime) <= pl.datetime(2023, 10, 20, 8, 35, 00)))
    print(f"Period: 2023-10-20 08:05 thru 2023-10-20 08:35")
    print("# ZERO Full Scatter")
    print(f"- Blue mean: {_df['BsB_S11'].mean()}")
    print(f"- Blue std: {_df['BsB_S11'].std()}")
    print(f"- Green mean: {_df['BsG_S11'].mean()}")
    print(f"- Green std: {_df['BsG_S11'].std()}")
    print(f"- Red mean: {_df['BsR_S11'].mean()}")
    print(f"- Red std: {_df['BsR_S11'].std()}")
    print("# ZERO Backcatter")
    print(f"- Blue mean: {_df['BbsB_S11'].mean()}")
    print(f"- Blue std: {_df['BbsB_S11'].std()}")
    print(f"- Green mean: {_df['BbsG_S11'].mean()}")
    print(f"- Green std: {_df['BbsG_S11'].std()}")
    print(f"- Red mean: {_df['BbsR_S11'].mean()}")
    print(f"- Red std: {_df['BbsR_S11'].std()}")
    print()
    return None

# span check
def print_span_check_results(df):
    _df = df.filter((pl.col("DateTime").cast(pl.Datetime) >= pl.datetime(2023, 10, 20, 11, 30, 00))
                    & (pl.col("DateTime").cast(pl.Datetime) <= pl.datetime(2023, 10, 20, 12, 25, 00)))
    print(f"Period: 2023-10-20 11:30 thru 2023-10-20 12:25")
    print("# CO2 Span Full Scatter")
    print(f"- Blue mean: {_df['BsB_S11'].mean()} (expected: 44.21)")
    print(f"- Blue std: {_df['BsB_S11'].std()}")
    print(f"- Green mean: {_df['BsG_S11'].mean()} (expected: 23.86)")
    print(f"- Green std: {_df['BsG_S11'].std()}")
    print(f"- Red mean: {_df['BsR_S11'].mean()} (expected: 11.15)")
    print(f"- Red std: {_df['BsR_S11'].std()}")
    print("# CO2 Span Backscatter")
    print(f"- Blue mean: {_df['BbsB_S11'].mean()} (expected: 22.11)")
    print(f"- Blue std: {_df['BbsB_S11'].std()}")
    print(f"- Green mean: {_df['BbsG_S11'].mean()} (expected: 11.93)")
    print(f"- Green std: {_df['BbsG_S11'].std()}")
    print(f"- Red mean: {_df['BbsR_S11'].mean()} (expected: 5.58)")
    print(f"- Red std: {_df['BbsR_S11'].std()}")
    print()
    return None
```

```python
df = pl.read_parquet("/home/zue/users/jkl/Public/git/gawkenya/results/aerosol/2023/10/S11a.parquet")
print_zero_check_results(df)
print_span_check_results(df)
```

```python
import os
import datetime
import yaml

from monitoring import kplc

with open("secrets-kplc", 'r') as fh:
    secrets = yaml.load(fh, Loader=yaml.FullLoader)

base_url = secrets['base_url']
usr = str(secrets['usr'])
pwd = secrets['pwd']

# Not sure if the following will work indeendent of platform!
root = os.getcwd()

target = os.path.join(os.path.expanduser(root), "results/kplc")

from_days_ago = 21
```

```python
to_date = datetime.datetime.today()
from_date = (to_date - datetime.timedelta(days=from_days_ago)).strftime("%Y-%m-%d")
to_date = to_date.strftime("%Y-%m-%d")

try:
    df = kplc.download_kplc_smartmeter_load_profile(usr=usr, pwd=pwd, from_date=from_date, to_date=to_date, base_url=base_url, target=target)
    if not(df.empty):
        print(df.tail())

    # if not df.empty:
    #     file = os.path.join(target, f"load_profile-{to_date.replace('-', '')}.csv")
    #     df.to_csv(file)

    #     df = kplc.read_kplc_smartmeter_load_profile(file, fix=True, save=True)

        kplc.plot_kplc_smartmeter_load_profile(df, target)

except Exception as err:
    print(err)
```

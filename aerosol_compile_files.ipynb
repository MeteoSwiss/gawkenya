```python
import os
from processing.cpd2 import CPD2
from processing.ae33 import AE33
```

```python
# process CPD2 data files
cpd2 = CPD2()
years = ["2016", "2017", "2020", "2021", "2022", "2023"]
for year in years:
    source = os.path.join("/product_data/data/pay/Kenya/MKN/incoming/aerosol", year)
    target = os.path.join("data", "level1", year)
    errors = cpd2.tarballs_to_parquet(source=source, target=target)
    print(errors)
print("done")
```

```python
# process AE33 data files
ae33 = AE33()
years = ["2022", "2023"]
for year in years:
    source = os.path.join("/product_data/data/pay/Kenya/MKN/incoming/ae33/data", year)
    target = os.path.join("data", "level1", year)
    errors = ae33.zipfiles_to_parquet(source=source, target=target)
    print(errors)
print("done")
```
